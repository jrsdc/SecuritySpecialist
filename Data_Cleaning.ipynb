<p>Importing a raw data set so I can dsiplay CRISP-DM methodology while cleaning this data set before designing the dashboard.

```python
#Imported a Generated Raw Dataset

import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# 1. Configuration
num_records = 1500
events = [
    "CCTV Motion Alert", "Intruder Alarm", "Access Denied", "Door Forced",
    "Suspicious Behaviour", "Perimeter Breach", "Fire Alarm", "Officer Welfare Check"
]
facilities = ["HQ_Main_Entrance", "Data_Center_PodA", "Warehouse_Gate4", "Server_Room_3B", "Loading_Dock"]

# 2. Generate Base Data
data = []
base_time = datetime(2026, 5, 1)

for i in range(num_records):
    timestamp = base_time + timedelta(minutes=random.randint(1, 40000), seconds=random.randint(0, 59))
    event = random.choice(events)
    location = random.choice(facilities)
    device_id = f"DEV-{random.randint(100, 999)}"
    severity = random.choice(["Low", "Medium", "High", "Critical"])

    # Introduce messy trailing/leading spaces to strings
    if random.random() < 0.15:
        event = f"  {event} "
    if random.random() < 0.10:
        location = f"{location}   "

    data.append([timestamp, device_id, event, location, severity])

df = pd.DataFrame(data, columns=["Timestamp", "Device_ID", "Event_Type", "Location", "Severity"])

# 3. Inject Deliberate "Dirty Data" for Cleaning
# Inject malformed dates (convert some datetimes to inconsistent string formats)
df['Timestamp'] = df['Timestamp'].astype(str)
mask_date = np.random.rand(len(df)) < 0.05
df.loc[mask_date, 'Timestamp'] = "00-00-2026 00:00"

# Inject Missing Values (NaNs)
mask_null_event = np.random.rand(len(df)) < 0.04
mask_null_loc = np.random.rand(len(df)) < 0.03
df.loc[mask_null_event, 'Event_Type'] = np.nan
df.loc[mask_null_loc, 'Location'] = np.nan

# Inject Exact Duplicate Rows
duplicates = df.sample(n=45)
df = pd.concat([df, duplicates], ignore_index=True)

# Shuffle and Save
df = df.sample(frac=1).reset_index(drop=True)
df.to_csv("raw_physical_security_logs.csv", index=False)
print("Data generated! File saved as 'raw_physical_security_logs.csv'")
```
